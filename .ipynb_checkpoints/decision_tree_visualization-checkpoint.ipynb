```python
from dtreeviz import model
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load dataset
iris = load_iris()
X = iris.data
y = iris.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

# Train model
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Create visualization model (NEW API)
viz_model = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

# Display tree
viz = viz_model.view()
viz
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import fetch_california_housing
from dtreeviz import model

# Load dataset
data = fetch_california_housing()

X_train = data.data
y_train = data.target

# Train model
regr = DecisionTreeRegressor(max_depth=1)
regr.fit(X_train, y_train)

# dtreeviz v2 syntax
viz_model = model(
    regr,
    X_train=X_train,
    y_train=y_train,
    feature_names=data.feature_names,
    target_name="price"
)

viz = viz_model.view()
viz
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from dtreeviz import model

# Load dataset
iris = load_iris()
X_train = iris.data
y_train = iris.target

# Train classifier
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Pick random sample to highlight path
X_instance = X_train[np.random.randint(0, len(X_train))]

# Create dtreeviz model (NEW v2 API)
viz_model = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

# View tree and highlight prediction path
viz = viz_model.view(x=X_instance)
viz
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from dtreeviz import model

# Load dataset
iris = load_iris()
X_train = iris.data
y_train = iris.target

# Train classifier
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Create visualization model (NEW API)
viz_model = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

# View tree with your custom settings
viz = viz_model.view(
    orientation="LR",          # Left → Right
    scale=1.5,                 # Zoom size
    show_node_labels=True      # Show node IDs
)

viz
```

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_iris
from dtreeviz import model

# Load dataset
iris = load_iris()
X_train = iris.data
y_train = iris.target

# Train classifier
clas = DecisionTreeClassifier()
clas.fit(X_train, y_train)

# Select random instance
X_instance = X_train[np.random.randint(0, len(X_train))]

# Create dtreeviz model
viz_model = model(
    clas,
    X_train=X_train,
    y_train=y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

# Show ONLY prediction path
viz = viz_model.view(
    x=X_instance,
    show_just_path=True
)

viz
```
